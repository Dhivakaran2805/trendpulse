```python
def load_country(code, data_dir=DATA_DIR):
    """Load one country's CSV and return a list of dicts (one per row)."""
    path = f"{data_dir}/{code}videos.csv"
    try:
        df = pd.read_csv(path)
    except UnicodeDecodeError:
        # Some files contain characters UTF-8 can't read
        df = pd.read_csv(path, encoding="latin-1")

    records = df.to_dict("records")
    for row in records:
        row["country"] = code      # remember where each row came from
    return records
```
